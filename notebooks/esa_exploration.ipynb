# ESA Headward Fluid Shift: Exploratory Data Analysis
**Study:** Determining the Dose Response Profile of the Headward Fluid Shift during Varying Gravity Levels (PFC82, DOI: 10.57780/esa-0ggni9p)


```python
import sys
sys.path.insert(0, "../src")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from vascushift.data.esa_loader import load_esa_data

df, report, meta = load_esa_data(data_dir="../data")
print(report.summary())
df.head(10)
```

```python
# Gravitational dose response curve for IJV area
plt.figure(figsize=(8, 5))
df.groupby("gz")["IJV_area_mm2"].agg(["mean", "std"]).plot(y="mean", yerr="std", capsize=4, marker="o", title="IJV Area vs Gz")
plt.ylabel("IJV Lumen Area (mm2)")
plt.xlabel("Gravitational Load Gz")
plt.grid(True)
plt.show()
```
